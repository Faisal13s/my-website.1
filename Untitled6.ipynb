```python
import pandas as pd
import numpy as np

sales = pd.read_csv(
    "sales_messy.csv",
    na_values=["NA", "N/A", "?", "null", "None", ""]
)

sales.head()
```

```python

sales.isna().sum()
```

```python
for col in sales.columns:
    print("\n---", col, "---")
    print(sales[col].value_counts(dropna=False))
```

```python
# تنظيف أسماء المدن
sales["city"] = sales["city"].str.strip().str.lower()

sales["city"] = sales["city"].replace({
    "riyadh": "Riyadh",
    "dammam": "Dammam",
    "jeddah": "Jeddah",
    "jiddah": "Jeddah"
})


# تنظيف طرق الدفع
sales["payment_type"] = sales["payment_type"].str.strip().str.lower()

sales["payment_type"] = sales["payment_type"].replace({
    "credit": "Credit Card",
    "credit card": "Credit Card",
    "creditcard": "Credit Card",
    "debit": "Debit",
    "cash": "Cash"
})


# تحويل السعر والكمية إلى أرقام
sales["price"] = pd.to_numeric(sales["price"], errors="coerce")
sales["quantity"] = pd.to_numeric(sales["quantity"], errors="coerce")


# تحويل التاريخ
sales["date"] = pd.to_datetime(sales["date"], errors="coerce")


# تعويض القيم المفقودة في السعر والكمية بالـ Median
sales["price"] = sales["price"].fillna(sales["price"].median())
sales["quantity"] = sales["quantity"].fillna(sales["quantity"].median())


# القيم النصية المفقودة
sales["customer_name"] = sales["customer_name"].fillna("Unknown")
sales["payment_type"] = sales["payment_type"].fillna("Unknown")


# عرض البيانات بعد التنظيف
sales
```
